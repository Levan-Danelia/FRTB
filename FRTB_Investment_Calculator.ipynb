## 1. Import Libraries

```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import warnings
warnings.filterwarnings('ignore')

print("Libraries imported successfully!")
```

## 2. Investment Calculator Functions

```python
def future_value(principal, rate, periods, compounding='annual'):
    """
    Calculate future value of an investment.
    
    Parameters:
    -----------
    principal : float
        Initial investment amount
    rate : float
        Annual interest rate (as decimal, e.g., 0.05 for 5%)
    periods : int
        Number of periods (years, months, etc.)
    compounding : str
        Compounding frequency: 'annual', 'semi-annual', 'quarterly', 'monthly', 'daily', 'continuous'
    
    Returns:
    --------
    float : Future value of investment
    """
    compounding_periods = {
        'annual': 1,
        'semi-annual': 2,
        'quarterly': 4,
        'monthly': 12,
        'daily': 365,
        'continuous': None
    }
    
    if compounding == 'continuous':
        return principal * np.exp(rate * periods)
    else:
        n = compounding_periods.get(compounding, 1)
        return principal * (1 + rate/n)**(n * periods)


def present_value(future_value, rate, periods, compounding='annual'):
    """
    Calculate present value of a future amount.
    
    Parameters:
    -----------
    future_value : float
        Future value amount
    rate : float
        Discount rate (as decimal)
    periods : int
        Number of periods
    compounding : str
        Compounding frequency
    
    Returns:
    --------
    float : Present value
    """
    compounding_periods = {
        'annual': 1,
        'semi-annual': 2,
        'quarterly': 4,
        'monthly': 12,
        'daily': 365,
        'continuous': None
    }
    
    if compounding == 'continuous':
        return future_value / np.exp(rate * periods)
    else:
        n = compounding_periods.get(compounding, 1)
        return future_value / (1 + rate/n)**(n * periods)


def compound_interest(principal, rate, periods, additional_contribution=0, contribution_frequency='annual'):
    """
    Calculate compound interest with optional regular contributions.
    
    Parameters:
    -----------
    principal : float
        Initial investment
    rate : float
        Annual interest rate (as decimal)
    periods : int
        Number of years
    additional_contribution : float
        Regular additional contribution amount
    contribution_frequency : str
        Frequency of contributions: 'annual', 'monthly'
    
    Returns:
    --------
    dict : Dictionary with total value, total contributions, and interest earned
    """
    if contribution_frequency == 'monthly':
        n = 12
        r = rate / 12
        t = periods * 12
    else:  # annual
        n = 1
        r = rate
        t = periods
    
    # Future value of initial principal
    fv_principal = principal * (1 + r)**t
    
    # Future value of annuity (regular contributions)
    if additional_contribution > 0:
        fv_contributions = additional_contribution * (((1 + r)**t - 1) / r)
    else:
        fv_contributions = 0
    
    total_value = fv_principal + fv_contributions
    total_contributions = principal + (additional_contribution * t)
    interest_earned = total_value - total_contributions
    
    return {
        'total_value': total_value,
        'total_contributions': total_contributions,
        'interest_earned': interest_earned,
        'return_percentage': (interest_earned / total_contributions) * 100
    }


def calculate_roi(initial_investment, final_value, holding_period_years=None):
    """
    Calculate Return on Investment (ROI).
    
    Parameters:
    -----------
    initial_investment : float
        Initial investment amount
    final_value : float
        Final value of investment
    holding_period_years : float, optional
        Holding period in years (for annualized ROI)
    
    Returns:
    --------
    dict : Dictionary with ROI metrics
    """
    gain = final_value - initial_investment
    roi = (gain / initial_investment) * 100
    
    result = {
        'gain': gain,
        'roi_percentage': roi
    }
    
    if holding_period_years:
        annualized_roi = ((final_value / initial_investment)**(1/holding_period_years) - 1) * 100
        result['annualized_roi_percentage'] = annualized_roi
    
    return result


def annuity_future_value(payment, rate, periods):
    """
    Calculate future value of an annuity.
    
    Parameters:
    -----------
    payment : float
        Regular payment amount
    rate : float
        Interest rate per period (as decimal)
    periods : int
        Number of payment periods
    
    Returns:
    --------
    float : Future value of annuity
    """
    return payment * (((1 + rate)**periods - 1) / rate)


def annuity_present_value(payment, rate, periods):
    """
    Calculate present value of an annuity.
    
    Parameters:
    -----------
    payment : float
        Regular payment amount
    rate : float
        Interest rate per period (as decimal)
    periods : int
        Number of payment periods
    
    Returns:
    --------
    float : Present value of annuity
    """
    return payment * ((1 - (1 + rate)**(-periods)) / rate)


def portfolio_analysis(assets_df):
    """
    Analyze investment portfolio.
    
    Parameters:
    -----------
    assets_df : pandas.DataFrame
        DataFrame with columns: 'asset_name', 'initial_value', 'current_value', 'weight'
    
    Returns:
    --------
    dict : Portfolio analysis results
    """
    total_initial = assets_df['initial_value'].sum()
    total_current = assets_df['current_value'].sum()
    
    assets_df['gain_loss'] = assets_df['current_value'] - assets_df['initial_value']
    assets_df['return_pct'] = (assets_df['gain_loss'] / assets_df['initial_value']) * 100
    
    portfolio_return = ((total_current - total_initial) / total_initial) * 100
    
    return {
        'total_initial_value': total_initial,
        'total_current_value': total_current,
        'total_gain_loss': total_current - total_initial,
        'portfolio_return_pct': portfolio_return,
        'assets_detail': assets_df
    }


print("Investment calculator functions defined successfully!")
```

## 3. Example: Future Value Calculation

```python
# Calculate future value of $10,000 invested at 7% annual rate for 10 years
principal = 10000
rate = 0.07
years = 10

fv_annual = future_value(principal, rate, years, 'annual')
fv_monthly = future_value(principal, rate, years, 'monthly')
fv_continuous = future_value(principal, rate, years, 'continuous')

print("=" * 60)
print("FUTURE VALUE CALCULATION")
print("=" * 60)
print(f"Principal: ${principal:,.2f}")
print(f"Annual Rate: {rate*100}%")
print(f"Time Period: {years} years")
print("-" * 60)
print(f"FV (Annual Compounding):     ${fv_annual:,.2f}")
print(f"FV (Monthly Compounding):    ${fv_monthly:,.2f}")
print(f"FV (Continuous Compounding): ${fv_continuous:,.2f}")
print("=" * 60)
```

## 4. Example: Compound Interest with Regular Contributions

```python
# Calculate compound interest with monthly contributions
initial_investment = 5000
annual_rate = 0.08
years = 20
monthly_contribution = 200

result = compound_interest(initial_investment, annual_rate, years, 
                          monthly_contribution, 'monthly')

print("=" * 60)
print("COMPOUND INTEREST WITH REGULAR CONTRIBUTIONS")
print("=" * 60)
print(f"Initial Investment: ${initial_investment:,.2f}")
print(f"Annual Rate: {annual_rate*100}%")
print(f"Time Period: {years} years")
print(f"Monthly Contribution: ${monthly_contribution:,.2f}")
print("-" * 60)
print(f"Total Value:         ${result['total_value']:,.2f}")
print(f"Total Contributions: ${result['total_contributions']:,.2f}")
print(f"Interest Earned:     ${result['interest_earned']:,.2f}")
print(f"Return:              {result['return_percentage']:.2f}%")
print("=" * 60)
```

## 5. Example: ROI Calculation

```python
# Calculate ROI for an investment
initial = 15000
final = 22500
holding_years = 3

roi_result = calculate_roi(initial, final, holding_years)

print("=" * 60)
print("RETURN ON INVESTMENT (ROI) CALCULATION")
print("=" * 60)
print(f"Initial Investment: ${initial:,.2f}")
print(f"Final Value:        ${final:,.2f}")
print(f"Holding Period:     {holding_years} years")
print("-" * 60)
print(f"Gain:               ${roi_result['gain']:,.2f}")
print(f"ROI:                {roi_result['roi_percentage']:.2f}%")
print(f"Annualized ROI:     {roi_result['annualized_roi_percentage']:.2f}%")
print("=" * 60)
```

## 6. Example: Portfolio Analysis

```python
# Create sample portfolio
portfolio_data = {
    'asset_name': ['Stocks', 'Bonds', 'Real Estate', 'Commodities', 'Cash'],
    'initial_value': [50000, 30000, 40000, 15000, 10000],
    'current_value': [62000, 31500, 48000, 13500, 10200],
    'weight': [0.35, 0.20, 0.28, 0.10, 0.07]
}

portfolio_df = pd.DataFrame(portfolio_data)
analysis = portfolio_analysis(portfolio_df)

print("=" * 80)
print("PORTFOLIO ANALYSIS")
print("=" * 80)
print(f"Total Initial Value:  ${analysis['total_initial_value']:,.2f}")
print(f"Total Current Value:  ${analysis['total_current_value']:,.2f}")
print(f"Total Gain/Loss:      ${analysis['total_gain_loss']:,.2f}")
print(f"Portfolio Return:     {analysis['portfolio_return_pct']:.2f}%")
print("-" * 80)
print("\nAsset Details:")
print(analysis['assets_detail'].to_string(index=False))
print("=" * 80)
```

## 7. Example: Annuity Calculations

```python
# Calculate annuity values
payment = 1000
rate_per_period = 0.06
periods = 15

fv_annuity = annuity_future_value(payment, rate_per_period, periods)
pv_annuity = annuity_present_value(payment, rate_per_period, periods)

print("=" * 60)
print("ANNUITY CALCULATIONS")
print("=" * 60)
print(f"Payment per Period: ${payment:,.2f}")
print(f"Rate per Period:    {rate_per_period*100}%")
print(f"Number of Periods:  {periods}")
print("-" * 60)
print(f"Future Value:       ${fv_annuity:,.2f}")
print(f"Present Value:      ${pv_annuity:,.2f}")
print("=" * 60)
```
